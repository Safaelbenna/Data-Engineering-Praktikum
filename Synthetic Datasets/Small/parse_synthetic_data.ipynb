```python
#parsing data and computing delta with the naive approach

data_graph_v1 = Graph()
data_graph_v2 = Graph()
data_graph_v1.parse("small_v1.ttl", format= "turtle")
data_graph_v2.parse("small_v2.ttl", format= "turtle")
print ("v1:")
for triple in data_graph_v1:
    print(triple)

print ("v2:")
for triple in data_graph_v2:
    print(triple) 

added = data_graph_v2 - data_graph_v1
print("added : ")
for triple in added:
    print(triple)  
print(len) 
 

```

```python
#computing delta with the hash based comparison algorithm

#version 1
hashes_v1 = set()
with open("small_v1.ttl", "r", encoding="utf-8") as f:
    for line in f:
        fingerprint = hashlib.sha256(line.strip().encode()).hexdigest()
        hashes_v1.add(fingerprint)
print("v1: ", hashes_v1)    

#version 2
hashes_v2 = set()
with open("small_v2.ttl", "r", encoding="utf-8") as f:
    for line in f:
        fingerprint = hashlib.sha256(line.strip().encode()).hexdigest()
        hashes_v2.add(fingerprint)
print("v2: ", hashes_v2)   

#changes
added = hashes_v2 - hashes_v1
print(len(added))
print("added: ", added)
```
