```python
from rdflib import Graph
```

```python
data_graph_v1 = Graph()
data_graph_v2 = Graph()
data_graph_v1.parse("small_v1.ttl", format= "turtle")
data_graph_v2.parse("small_v2.ttl", format= "turtle")
print ("v1:")
for triple in data_graph_v1:
    print(triple)



print ("v2:")

for triple in data_graph_v2:
    print(triple) 

added = data_graph_v2 - data_graph_v1
print("added : ")
for triple in added:
    print(triple)  

```
